# 📊 Unemployment Analysis with Python

### Objective
To analyze unemployment trends in India using Python, explore regional
patterns, visualize unemployment rates, and investigate changes during the
COVID-19 period.

### Tools Used
Python, Pandas, NumPy, Matplotlib, Seaborn

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df1 = pd.read_csv("Unemployment in India.csv")

print("First dataset loaded successfully!")
df1.head()
```

```python
df2 = pd.read_csv("Unemployment_Rate_upto_11_2020.csv")

print("Second dataset loaded successfully!")
df2.head()
```

```python
print("First Dataset Shape:", df1.shape)
print("Second Dataset Shape:", df2.shape)
```

```python
print("First Dataset Columns:")
print(df1.columns.tolist())

print("\nSecond Dataset Columns:")
print(df2.columns.tolist())
```

```python
print("First Dataset Information:")
df1.info()

print("\nSecond Dataset Information:")
df2.info()
```

```python
print("First Dataset Summary:")
display(df1.describe())

print("\nSecond Dataset Summary:")
display(df2.describe())
```

```python
print("Missing Values in First Dataset:")
print(df1.isnull().sum())

print("\nMissing Values in Second Dataset:")
print(df2.isnull().sum())
```

```python
print("Duplicate rows in First Dataset:", df1.duplicated().sum())
print("Duplicate rows in Second Dataset:", df2.duplicated().sum())
```

```python
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

print("Column names cleaned successfully!")
```

```python
print("First Dataset Columns:")
print(df1.columns.tolist())

print("\nSecond Dataset Columns:")
print(df2.columns.tolist())
```

```python
# Convert date columns into datetime format
for df in [df1, df2]:
    for col in df.columns:
        if "Date" in col:
            df[col] = pd.to_datetime(
                df[col],
                dayfirst=True,
                errors="coerce"
            )

print("Date conversion completed!")
```

```python
print(df1.dtypes)
print("\n-----------------------------\n")
print(df2.dtypes)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df1,
    x="Estimated Unemployment Rate (%)",
    kde=True
)

plt.title("Distribution of Estimated Unemployment Rate")
plt.xlabel("Unemployment Rate (%)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=df1,
    x="Date",
    y="Estimated Unemployment Rate (%)"
)

plt.title("Unemployment Rate Trend in India")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")

plt.xticks(rotation=45)
plt.show()
```

```python
state_unemployment = (
    df1.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
)

state_unemployment.head(10)
```

```python
plt.figure(figsize=(12, 6))

state_unemployment.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Regions by Average Unemployment Rate")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=df1,
    x="Date",
    y="Estimated Employed"
)

plt.title("Estimated Employment Trend in India")
plt.xlabel("Date")
plt.ylabel("Estimated Employed")

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=df1,
    x="Date",
    y="Estimated Labour Participation Rate (%)"
)

plt.title("Labour Participation Rate Over Time")
plt.xlabel("Date")
plt.ylabel("Labour Participation Rate (%)")

plt.xticks(rotation=45)
plt.show()
```

```python
covid_period = df1[
    (df1["Date"] >= "2020-03-01") &
    (df1["Date"] <= "2020-06-30")
]

print("COVID-19 period records:")
display(covid_period.head())

print("\nAverage Unemployment Rate during March-June 2020:")
print(
    covid_period["Estimated Unemployment Rate (%)"].mean()
)
```

```python
plt.figure(figsize=(10, 6))

sns.lineplot(
    data=covid_period,
    x="Date",
    y="Estimated Unemployment Rate (%)",
    marker="o"
)

plt.title("Unemployment Rate During March-June 2020")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")

plt.xticks(rotation=45)
plt.show()
```

```python
pre_covid = df1[
    df1["Date"] < "2020-03-01"
]["Estimated Unemployment Rate (%)"].mean()

covid = df1[
    (df1["Date"] >= "2020-03-01") &
    (df1["Date"] <= "2020-06-30")
]["Estimated Unemployment Rate (%)"].mean()

comparison = pd.DataFrame({
    "Period": ["Pre-COVID", "COVID-19 Period"],
    "Average Unemployment Rate": [pre_covid, covid]
})

comparison
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(
    data=comparison,
    x="Period",
    y="Average Unemployment Rate"
)

plt.title("Pre-COVID vs COVID-19 Unemployment Rate")
plt.xlabel("Period")
plt.ylabel("Average Unemployment Rate (%)")

plt.show()
```

```python
covid_state_analysis = (
    covid_period.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
)

covid_state_analysis.head(10)
```

```python
plt.figure(figsize=(12, 6))

covid_state_analysis.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Regions by Unemployment Rate During COVID-19")
plt.xlabel("Average Unemployment Rate (%)")
plt.ylabel("Region")

plt.show()
```

```python
overall_avg = df1["Estimated Unemployment Rate (%)"].mean()
highest_rate = df1["Estimated Unemployment Rate (%)"].max()
lowest_rate = df1["Estimated Unemployment Rate (%)"].min()

print("Overall Average Unemployment Rate:", round(overall_avg, 2), "%")
print("Highest Recorded Unemployment Rate:", round(highest_rate, 2), "%")
print("Lowest Recorded Unemployment Rate:", round(lowest_rate, 2), "%")
```

```python
print("========== KEY INSIGHTS ==========")

print(f"1. Overall average unemployment rate: {overall_avg:.2f}%")
print(f"2. Highest recorded unemployment rate: {highest_rate:.2f}%")
print(f"3. Lowest recorded unemployment rate: {lowest_rate:.2f}%")
print(f"4. Average unemployment rate during the COVID-19 period: {covid:.2f}%")
print(f"5. Average unemployment rate before the COVID-19 period: {pre_covid:.2f}%")
```
